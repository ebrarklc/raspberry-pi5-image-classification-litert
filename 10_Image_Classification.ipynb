```python
from ai_edge_litert.interpreter import Interpreter

model_path = "./models/mobilenet_v2_1.0_224_quant.tflite"
interpreter = Interpreter(model_path=model_path)
interpreter.allocate_tensors()

input_details = interpreter.get_input_details()
output_details = interpreter.get_output_details()

print("GİRİŞ :", input_details[0]['shape'], input_details[0]['dtype'],
      "| quantization:", input_details[0]['quantization'])
print("ÇIKIŞ :", output_details[0]['shape'], output_details[0]['dtype'],
      "| quantization:", output_details[0]['quantization'])
```

```python
img = Image.open("Cat03.jpg").convert("RGB")
h, w = input_details[0]['shape'][1], input_details[0]['shape'][2]
img_resized = img.resize((w, h))
input_data = np.expand_dims(np.array(img_resized), axis=0)
print("Model girişi:", input_data.shape, input_data.dtype)
```

```python
interpreter.set_tensor(input_details[0]['index'], input_data)

# Isınma turu (warm-up), Bölüm 1'deki benchmark kuralı!
interpreter.invoke()

times = []
for _ in range(20):
    t0 = time.perf_counter()
    interpreter.invoke()
    times.append((time.perf_counter() - t0) * 1000)

print(f"Çıkarım süresi: {np.mean(times):.1f} ± {np.std(times):.1f} ms  (20 tekrar)")
predictions = interpreter.get_tensor(output_details[0]['index'])[0]
print("Ham çıktı:", predictions.shape, predictions.dtype, "| ilk 10:", predictions[:10])
```

```python
def load_labels(filename):
    with open(filename, 'r') as f:
        return [line.strip() for line in f.readlines()]

labels = load_labels("./models/labels.txt")
top_k_indices = np.argsort(predictions)[::-1][:5]

for i in top_k_indices:
    print(f"{i:4d}  {labels[i]:20s}  ham değer: {predictions[i]}")
```

```python
scale, zero_point = output_details[0]['quantization']
dequantized = (predictions.astype(np.float32) - zero_point) * scale

print("Min:", dequantized.min(), "| Max:", dequantized.max(), "| Toplam:", dequantized.sum())
```

```python
exp_output = np.exp(dequantized - np.max(dequantized))
probabilities = exp_output / np.sum(exp_output)

print("Toplam olasılık:", probabilities.sum())
print()
for i in top_k_indices:
    print(f"  {labels[i]:20s}: {probabilities[i]*100:5.1f}%")
```

```python
for n in [1, 2, 4]:
    itp = Interpreter(model_path=model_path, num_threads=n)
    itp.allocate_tensors()
    itp.set_tensor(itp.get_input_details()[0]['index'], input_data)
    itp.invoke()  # warm-up
    t = []
    for _ in range(20):
        t0 = time.perf_counter()
        itp.invoke()
        t.append((time.perf_counter() - t0) * 1000)
    print(f"num_threads={n}: {np.mean(t):5.1f} ± {np.std(t):.1f} ms  →  ~{1000/np.mean(t):.0f} FPS")
```

```python
class ImageClassifier:
    def __init__(self, model_path, labels_path, num_threads=4):
        # Pahalı işler: SADECE BİR KEZ yapılır
        self.interpreter = Interpreter(model_path=model_path, num_threads=num_threads)
        self.interpreter.allocate_tensors()
        self.inp = self.interpreter.get_input_details()[0]
        self.out = self.interpreter.get_output_details()[0]
        self.h, self.w = self.inp['shape'][1], self.inp['shape'][2]
        self.scale, self.zero_point = self.out['quantization']
        with open(labels_path) as f:
            self.labels = [line.strip() for line in f]

    def classify(self, img, top_k=5):
        t0 = time.perf_counter()
        # ① Ön işleme
        if not isinstance(img, Image.Image):
            img = Image.fromarray(img)
        x = np.expand_dims(np.array(img.convert("RGB").resize((self.w, self.h))), axis=0)
        t1 = time.perf_counter()
        # ② Çıkarım
        self.interpreter.set_tensor(self.inp['index'], x)
        self.interpreter.invoke()
        raw = self.interpreter.get_tensor(self.out['index'])[0]
        t2 = time.perf_counter()
        # ③ Son işleme: dequantize + softmax + top-k
        logits = (raw.astype(np.float32) - self.zero_point) * self.scale
        e = np.exp(logits - logits.max())
        probs = e / e.sum()
        top = np.argsort(probs)[::-1][:top_k]
        t3 = time.perf_counter()

        results = [(self.labels[i], float(probs[i])) for i in top]
        times = {"pre": (t1-t0)*1000, "infer": (t2-t1)*1000, "post": (t3-t2)*1000}
        return results, times
```

```python
clf = ImageClassifier("./models/mobilenet_v2_1.0_224_quant.tflite", "./models/labels.txt")

results, times = clf.classify(Image.open("Cat03.jpg"))
for label, p in results:
    print(f"  {label:20s} {p*100:5.1f}%")
print(f"\n⏱️ Ön işleme: {times['pre']:.1f} ms | Çıkarım: {times['infer']:.1f} ms | Son işleme: {times['post']:.2f} ms")
```

```python
# Doküman yöntemi: her seferinde modeli yükle + çalıştır
t0 = time.perf_counter()
for _ in range(10):
    itp = Interpreter(model_path="./models/mobilenet_v2_1.0_224_quant.tflite", num_threads=4)
    itp.allocate_tensors()
    itp.set_tensor(itp.get_input_details()[0]['index'], input_data)
    itp.invoke()
naive = (time.perf_counter() - t0) * 100   # 10 tekrar → ms/görüntü

# Bizim yöntem: model bir kez yüklendi
t0 = time.perf_counter()
for _ in range(10):
    clf.classify(Image.open("Cat03.jpg"))
ours = (time.perf_counter() - t0) * 100

print(f"Her seferinde yükle : {naive:6.1f} ms/görüntü")
print(f"Bir kez yükle       : {ours:6.1f} ms/görüntü")
print(f"Kazanç              : {naive/ours:.1f}x")
```
